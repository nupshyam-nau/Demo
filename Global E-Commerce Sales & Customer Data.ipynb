```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")

# Convert date column
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

**1. Sales Trend Over Time**

```python
sales_time = df.groupby("Order_Date")["Total_Sales"].sum()

plt.figure(figsize=(10,5))
sns.lineplot(x=sales_time.index, y=sales_time.values)
plt.title("Sales Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

**2. Profit Trend Over Time**

```python
profit_time = df.groupby("Order_Date")["Profit"].sum()

plt.figure(figsize=(10,5))
sns.lineplot(x=profit_time.index, y=profit_time.values)
plt.title("Profit Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

**3. Sales by Customer Segment**

```python
segment_sales = df.groupby("Customer_Segment")["Total_Sales"].sum().reset_index()

plt.figure(figsize=(7,4))
sns.barplot(data=segment_sales,
            x="Customer_Segment",
            y="Total_Sales")
plt.title("Sales by Customer Segment")
plt.show()
```

**4. Profit by Customer Segment**

```python
segment_profit = df.groupby("Customer_Segment")["Profit"].sum().reset_index()

sns.barplot(data=segment_profit, x="Customer_Segment", y="Profit")
plt.title("Profit by Customer Segment")
plt.show()
```

**5. Sales by Region**

```python
region_sales = df.groupby("Region")["Total_Sales"].sum().reset_index()

sns.barplot(data=region_sales, x="Region", y="Total_Sales")
plt.title("Sales by Region")
plt.xticks(rotation=45)
plt.show()
```

**6. Top Countries by Sales**

```python
country_sales = (df.groupby("Country")["Total_Sales"].sum().nlargest(10).reset_index())
sns.barplot(data=country_sales, y="Country", x="Total_Sales")

plt.title("Top 10 Countries by Sales")
plt.show()
```

**7. Product Category Sales Analysis**

```python
cat_sales = df.groupby("Product_Category")["Total_Sales"].sum().reset_index()

sns.barplot(data=cat_sales, x="Product_Category", y="Total_Sales")

plt.title("Sales by Product Category")
plt.xticks(rotation=45)
plt.show()
```

**8. Product Category Profit Analysis**

```python
cat_profit = df.groupby("Product_Category")["Profit"].sum().reset_index()

sns.barplot(data=cat_profit, x="Product_Category", y="Profit")

plt.title("Profit by Product Category")
plt.xticks(rotation=45)
plt.show()
```

**10. Top 10 Products by Profit**

```python
top_profit_products = (df.groupby("Product_Name")["Profit"].sum().nlargest(10).reset_index())
sns.barplot(data=top_profit_products, y="Product_Name", x="Profit")

plt.title("Top 10 Products by Profit")
plt.show()
```

**11. Quantity vs Total Sales**

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x="Quantity", y="Total_Sales")

plt.title("Quantity vs Sales")
plt.show()
```

**12. Unit Price vs Total Sales**

```python
sns.scatterplot(data=df, x="Unit_Price", y="Total_Sales")

plt.title("Unit Price vs Sales")
plt.show()
```

**13. Discount Impact on Profit**

```python
sns.scatterplot(data=df, x="Discount_Percent", y="Profit")

plt.title("Discount vs Profit")
plt.show()
```

**14. Discount vs Sales**

```python
sns.scatterplot(data=df, x="Discount_Percent", y="Total_Sales")

plt.title("Discount vs Sales")
plt.show()
```

**15. Shipping Cost vs Profit**

```python
sns.scatterplot(data=df, x="Shipping_Cost", y="Profit")

plt.title("Shipping Cost vs Profit")
plt.show()
```

**16. Payment Method Distribution**

```python
payment = df["Payment_Method"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(payment.values, labels=payment.index, autopct="%1.1f%%")

plt.title("Payment Method Distribution")
plt.show()
```

**17. Customer Segment vs Payment Method**

```python
cross = pd.crosstab(df["Customer_Segment"], df["Payment_Method"])
cross.plot(kind="bar", stacked=True, figsize=(8,5))

plt.title("Segment vs Payment Method")
plt.show()
```

**18. Product Category vs Customer Segment Heatmap**

```python
heat = df.pivot_table(values="Total_Sales", index="Product_Category", columns="Customer_Segment", aggfunc="sum")
sns.heatmap(heat, annot=True, cmap="Blues")

plt.title("Category vs Customer Segment")
plt.show()
```

**19. Sales Distribution**

```python
sns.histplot(df["Total_Sales"], bins=30, kde=True)

plt.title("Sales Distribution")
plt.show()
```

**20. Profit Distribution**

```python
sns.histplot(df["Profit"], bins=30, kde=True)

plt.title("Profit Distribution")
plt.show()
```

**21. Correlation Heatmap**

```python
corr = df[["Quantity","Unit_Price","Discount_Percent","Total_Sales","Shipping_Cost","Profit"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

**22. Top Customers by Sales**

```python
customers = (df.groupby("Customer_Name")["Total_Sales"].sum().nlargest(10).reset_index())
sns.barplot(data=customers, y="Customer_Name", x="Total_Sales")

plt.title("Top Customers by Sales")
plt.show()
```

**23. Average Order Value by Customer Segment**

```python
aov = (df.groupby("Customer_Segment")["Total_Sales"].mean().reset_index())
sns.barplot(data=aov, x="Customer_Segment", y="Total_Sales")

plt.title("Average Order Value by Segment")
plt.show()
```
